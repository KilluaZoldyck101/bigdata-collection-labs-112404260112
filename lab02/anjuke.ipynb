```python
def parse_house(row):
    # 提取普通字段
    def get_text(selector):
        node = row.select_one(selector)
        return node.get_text(strip=True) if node else ""

    title = get_text("h3.property-content-title-name")
    house = get_text("p.property-content-info-comm-name")
    address = get_text("p.property-content-info-comm-address")
    price = get_text("p.property-price-average")

    # 获取房源属性列表
    info = row.select(
        "div.property-content-info > p.property-content-info-text"
    )

    values = [node.get_text(strip=True) for node in info]

    struct = values[0] if len(values) > 0 else ""
    area = values[1] if len(values) > 1 else ""
    floor = values[3] if len(values) > 3 else ""

    # 建造年代缺失时使用空字符串
    maketime = ""
    for value in values:
        if "年建造" in value:
            maketime = value
            break

    return [
        title, house, address, struct,
        area, floor, maketime, price
    ]


# 测试第一条房源
fields = [
    "title", "house", "address", "struct",
    "area", "floor", "maketime", "price"
]

if len(rows) > 0:
    first_house = parse_house(rows[0])

    for name, value in zip(fields, first_house):
        print(name, ":", value)
else:
    print("没有获取到房源数据")
```

```python
import time
from urllib.parse import urljoin, urlparse

# 保持之前成功使用的请求头
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/156.0.0.0 Safari/537.36 Edg/156.0.0.0"
}

current_url = "http://www.bspider.top/anjuke/"

all_data = []
visited = set()
max_pages = 30

# 记录采集是否正常完成
finished = False

for page in range(1, max_pages + 1):

    if current_url in visited:
        print("错误：页面地址重复，停止")
        break

    visited.add(current_url)

    try:
        r_page = requests.get(
            current_url,
            headers=headers,
            timeout=30
        )
        r_page.raise_for_status()
        r_page.encoding = r_page.apparent_encoding

        soup_page = BeautifulSoup(
            r_page.text, "html.parser"
        )

        # 检查是否进入验证页面
        title = (
            soup_page.title.get_text(strip=True)
            if soup_page.title else ""
        )

        if "访问验证" in title:
            print("错误：出现访问验证页面")
            break

        # 提取当前页房源
        page_rows = soup_page.select("div.property")

        if not page_rows:
            print("错误：本页没有找到房源")
            break

        page_data = [
            parse_house(row) for row in page_rows
        ]

        # 检查是否与前一页完全重复
        if page > 1 and page_data == previous_data:
            print("错误：页面内容重复")
            break

        all_data.extend(page_data)
        previous_data = page_data

        print(
            "第", page, "页",
            "本页：", len(page_data),
            "累计：", len(all_data),
            "URL：", current_url
        )

        # 找到分页区域
        pagination = soup_page.select_one(
            "section.pagination-wrap"
        )

        if pagination is None:
            print("错误：分页区域不存在")
            break

        # 找下一页按钮
        next_node = pagination.select_one(
            "a.next.next-active"
        )

        if next_node is None:
            # 检查是否有已禁用的下一页按钮
            disabled_next = pagination.select_one("a.next")

            if disabled_next is not None:
                print("已到达末页")
                finished = True
            else:
                print("错误：没有找到下一页按钮")

            break

        href = next_node.get("href")

        if not href:
            print("错误：下一页链接不存在")
            break

        next_url = urljoin(current_url, href)

        # 检查没有离开教学站
        parsed = urlparse(next_url)

        if (
            parsed.scheme not in ("http", "https")
            or parsed.hostname != "www.bspider.top"
            or not parsed.path.startswith("/anjuke/")
        ):
            print("错误：下一页地址不属于教学站")
            break

        if next_url in visited:
            print("错误：下一页地址已经访问过")
            break

        current_url = next_url

        # 每页之间至少等待1秒
        time.sleep(1)

    except requests.RequestException as e:
        print("请求失败：", e)
        break

if not finished:
    print("采集未确认正常到达末页，请检查日志")

print("采集到的原始房源总数：", len(all_data))
```

```python
import os
import pymysql

conn = None
cursor = None

try:
    conn = pymysql.connect(
        host="mysql",
        port=3306,
        user="student",
        password=os.environ["SMARTPARK_DB_PASSWORD"],
        database="smartpark_lab",
        charset="utf8mb4",
        autocommit=False
    )

    cursor = conn.cursor()

    cursor.execute("SELECT COUNT(*) FROM houseinfo")
    before_count = cursor.fetchone()[0]

    print("入库前数据条数：", before_count)

    cursor.execute("""
        SELECT title, house, address, struct,
               area, floor, maketime, price
        FROM houseinfo
    """)

    existing = set(cursor.fetchall())

    new_records = [
        record for record in records
        if record not in existing
    ]

    new_records = list(dict.fromkeys(new_records))

    print("CSV数据条数：", len(records))
    print("本次准备新增：", len(new_records))

    sql = """
        INSERT INTO houseinfo
        (title, house, address, struct,
         area, floor, maketime, price)
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
    """

    if new_records:
        affected = cursor.executemany(sql, new_records)
        conn.commit()
        print("本次实际插入：", affected)
    else:
        print("没有需要新增的数据")
        conn.commit()

    cursor.execute("SELECT COUNT(*) FROM houseinfo")
    after_count = cursor.fetchone()[0]

    print("入库后数据条数：", after_count)
    print("本次新增条数：", after_count - before_count)

except Exception as e:
    if conn is not None:
        conn.rollback()
    print("入库失败，已尝试回滚：", e)
    raise

finally:
    if cursor is not None:
        cursor.close()

    if conn is not None:
        conn.close()

    print("数据库连接已关闭")
```
